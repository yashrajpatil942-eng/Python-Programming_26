```python
matrix1=[1,5,10],[10,5,1]
matrix2=[3,4,6],[6,4,3]
resultmatrix=[0,0,0],[0,0,0]

for i in range(len(matrix1)):
    for j in range(len(matrix1[0])):

        resultmatrix[i][j]= matrix1[i][j] + matrix2[i][j]

print('matrix1 : ')
for row in matrix1:
    print(row)
print('matrix2 : ')
for roww in matrix2:
    print(roww)
print('resultmatrix : ')
for rowww in resultmatrix:
    print(rowww)
```
